```python
raw_sessions2 = raw_sessions1.copy(deep = True)
```

```python
total_sessions_table = raw_sessions2.groupby('user_id').size().reset_index(name = 'total_sessions')
total_sessions_table.head()
```

```python
total_sessions_table['user_id'].nunique()
```

```python
raw_sessions2['user_id'].nunique()
```

```python
booking_table = raw_sessions2[raw_sessions2['detail_group'] == 'booking_flow'].groupby('user_id').size().reset_index(name = 'booking_counts') 
booking_table['is_booking'] = (booking_table['booking_counts'] > 0).astype(int)
booking_table.head()
```

```python
booking_table['user_id'].nunique()
```

```python
new_sessions_table = total_sessions_table.merge(booking_table, how = 'left', on = 'user_id')
print('Current missing values before imputation:\n', new_sessions_table.isna().sum())
new_sessions_table = new_sessions_table.fillna(0)
print('\nCurrent missing values after imputation:\n', new_sessions_table.isna().sum())
print('\n', new_sessions_table.head())
print('\nThere are ', len(new_sessions_table), ' of observations.')
```

```python
abandon_list = ['guest_cancellation', 'delete_payment_instrument', 'create_alteration_request']
abandon_table = raw_sessions2[raw_sessions2['action_detail'].isin(abandon_list)].groupby('user_id').size().reset_index(name = 'abandon_freq') 
new_sessions_table = new_sessions_table.merge(abandon_table, how = 'left', on = 'user_id') 
new_sessions_table['regret_ratio'] = new_sessions_table['abandon_freq']/(new_sessions_table['booking_counts'] + 0.0001) 
print('Current missing values before imputation:\n', new_sessions_table.isna().sum())
new_sessions_table['regret_ratio'] = new_sessions_table['regret_ratio'].fillna(0)
new_sessions_table['abandon_freq'] = new_sessions_table['abandon_freq'].fillna(0).astype(int)
print('\nCurrent missing values after imputation:\n', new_sessions_table.isna().sum())
print('\n', new_sessions_table.head())
print('\nThere are ', len(new_sessions_table), ' of observations.')
```

```python
new_sessions_table['booking_intent_rate'] = new_sessions_table['booking_counts']/(new_sessions_table['total_sessions'] + 0.0001)
new_sessions_table['conversion_rate'] = (new_sessions_table['booking_counts'] - new_sessions_table['abandon_freq'])/new_sessions_table['total_sessions']
print('Current missing values:\n', new_sessions_table.isna().sum())
print('\n', new_sessions_table.head())
```

```python
# clicks before booking = total of view and search
total_searches_table = raw_sessions2[raw_sessions2['detail_group'] == 'search'].groupby('user_id').size().reset_index(name = 'total_searches')
total_engage_table = raw_sessions2[raw_sessions2['detail_group'] == 'engagement'].groupby('user_id').size().reset_index(name = 'total_engagement')
total_communicate_table = raw_sessions2[raw_sessions2['detail_group'] == 'communication'].groupby('user_id').size().reset_index(name = 'total_communication')
```

```python
total_searches_table.shape
```

```python
total_engage_table.shape
```

```python
total_communicate_table.shape
```

```python
clicks_before_booking_table = total_searches_table.merge(total_engage_table, how = 'outer', on = 'user_id')
```

```python
clicks_before_booking_table = clicks_before_booking_table.merge(total_communicate_table, how = 'outer', on = 'user_id')
clicks_before_booking_table.head()
```

```python
print('Current missing values before imputation:\n', clicks_before_booking_table.isna().sum())
clicks_before_booking_table = clicks_before_booking_table.fillna(0)
print('\nCurrent missing values after imputation:\n', clicks_before_booking_table.isna().sum())

clicks_before_booking_table['total_evaluation'] = clicks_before_booking_table['total_engagement'] + clicks_before_booking_table['total_communication']

new_sessions_table = new_sessions_table.merge(clicks_before_booking_table, how = 'left', on = 'user_id')
print('\nCurrent missing values before imputation:\n', new_sessions_table.isna().sum())
print('\n', new_sessions_table.head())
print('\nThere are ', len(new_sessions_table), ' of observations.')
```

```python
new_sessions_table = new_sessions_table.fillna(0)

new_sessions_table['decision_complexity'] = (new_sessions_table['total_searches'] + new_sessions_table['total_evaluation'])/(new_sessions_table['booking_counts'] + 0.0001)
print('Current missing values:\n', new_sessions_table.isna().sum())
print('\n', new_sessions_table.head())
print('\nThere are ', len(new_sessions_table), ' of observations.')
```

```python
new_sessions_table.loc[new_sessions_table['booking_counts'] == 0, 'decision_complexity'] = 0
new_sessions_table.head()
```

```python
new_sessions_table['has_regretted'] = (new_sessions_table['regret_ratio'] > 0).astype(int) 
new_sessions_table.head()
```

```python
# friction feature
friction_list = ['apply_coupon_error', 'at_checkpoint', 'profile_verifications', 
                 'account_payout_preferences', 'change_password', 'forgot_password']
friction_table = raw_sessions2[raw_sessions2['action_detail'].isin(friction_list)].groupby('user_id').size().reset_index(name = 'friction_freq')
friction_table['friction_freq'] = friction_table['friction_freq'].fillna(0)
friction_table.head()
```

```python
new_sessions_table = new_sessions_table.merge(friction_table, how = 'left', on = 'user_id')
print('Current missing values before imputation:\n', new_sessions_table.isna().sum())
new_sessions_table = new_sessions_table.fillna(0)
print('\nCurrent missing values after:\n', new_sessions_table.isna().sum())
print('\n', new_sessions_table.head())
print('\nThere are ', len(new_sessions_table), ' of observations.')
```

```python
browsing_table1 = raw_sessions2[raw_sessions2['action_detail'] == 'p3'].groupby('user_id').size().reset_index(name = 'p3_freq')
browsing_table2 = raw_sessions2[raw_sessions2['action_detail'] == 'view_search_results'].groupby('user_id').size().reset_index(name = 'view_search_freq')
browsing_table = browsing_table1.merge(browsing_table2, how = 'outer', on = 'user_id') 
browsing_table = browsing_table.fillna(0)
browsing_table['browsing_depth'] = browsing_table['p3_freq']/(browsing_table['view_search_freq'] + 0.0001)
browsing_table.head()
```
